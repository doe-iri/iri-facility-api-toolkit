```python
# Getting Started Example

There are a number of samples under the parent directory.

This basic tutorial is highly recommended to ensure your fabfed credential file 
is configured correctly for the fabric provider. 

Here we use a simple config [../getting_started/fabric](../getting_started/fabric/README.md) to create a couple of nodes 
in fabric on different Fabric sites. 

One of the nodes is the janus controller. At any time, you can set the count of janus dtn service to 1 and re-apply the config.
```

```python
import os
import shutil
import demo_helper as dh

# It is best practice to set your initials to name the session
initials = 'knit10'
session_name = f"{initials}-hello-fabric" 
experiment_path = '../exp_hello_fabric'
fab_file_path = '../getting_started/fabric/config.fab'
destination_file = f'{experiment_path}/config.fab'
    
if not os.path.exists(destination_file):
    os.mkdir(experiment_path)
    shutil.copy(fab_file_path, destination_file)
    print(f"The workflow config has been saved to {dh.absolute_path(experiment_path)}")
else:
    print(f"Found existing workflow config in {dh.absolute_path(experiment_path)}")

if not os.path.exists(dh.default_credentials_location()):
    print("WARNING: Please see the welcome notebook to prepare the credential file")
    
```

## Open Fabfed Workflow Definition
