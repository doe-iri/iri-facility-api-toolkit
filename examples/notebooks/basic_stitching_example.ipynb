```python
import os
import shutil
import demo_helper as dh

# It is best practice to add your initials to name the session
initials = "knit10"
session_name = f"{initials}-clab" 
experiment_path = '../exp_basic_stitching_cloudlab'
fab_file_path = '../basic-stitching/cloudlab/config.fab'
destination_file = f'{experiment_path}/config.fab'
    
if not os.path.exists(destination_file):
    os.mkdir(experiment_path)
    shutil.copy(fab_file_path, destination_file)
    print(f"The workflow config has been saved to {dh.absolute_path(experiment_path)}")
else:
    print(f"Found existing workflow config in {dh.absolute_path(experiment_path)}")

if not os.path.exists(dh.default_credentials_location()):
    print("WARNING: Please see the welcome notebook to prepare the credential file")
    
```
