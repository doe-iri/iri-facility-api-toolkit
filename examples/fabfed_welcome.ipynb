## Credential Configuration

The Fabfed credential file is a YAML file that contains account information and keying material for each provider using sections aka profiles. The code below will copy the [template file](./fabfed_config/fabfed_credentials_template.yml)  and link it to the standard location.

Note: Only providers that will be used need to be configured.

```python
import os
import shutil
import notebooks.demo_helper as dh

default_credentials = dh.default_credentials_location()
credentials = './exp_fabfed_credentials.yml'

if os.path.exists(default_credentials):
    print(f'WARNING: found existing default credential file:{default_credentials}')
else:
    os.makedirs('~/.fabfed', exist_ok=True)
    source_file = 'fabfed_config/fabfed_credentials_template.yml'
    shutil.copy(source_file, default_credentials)

if not os.path.exists(credentials):
    os.link(default_credentials, credentials)
    print(f'Created link to :{default_credentials}')
    
```
