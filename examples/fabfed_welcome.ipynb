## Janus DTN Service.

In addtion to nodes and stitched networks, Fabfed handles services (applications). Services are triggered only if the nodes and the networks have successfully been provisioned. The Janus service is bundled with fabfed and uses ansible under the hood to install docker and janus software on all the nodes. It consists of a single controller server node and one or more client nodes. 
The janus controller has a web-based ui that allows for creating janus sessions and launch dockerizedtools like iperf and xfer on the nodes listed in the janus session.   

- Note: The fabfed sample workflow definitions have janus service disabled by default. Thats is count is set to 0. You can set the count to 1 at anytime even after having applied the workflow.
- Note: The `fabfed workflow -s some_session -show -summary` dumps the command to create an ssh tunnel to the janus controller. The janus web-based ui can then be accessed locally at [http://localhost:8000](http://localhost:8000/)
